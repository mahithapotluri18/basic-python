```python
#DIRECT RECURSION(cALL)
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))
```

```python
#INDIRECT RECURSION
def is_even(n):
    if n == 0:
        return True
    else:
        return is_odd(n - 1)

def is_odd(n):
    if n == 0:
        return False
    else:
        return is_even(n - 1)

print(is_even(4))
print(is_odd(3))
```

```python
#super factorial
def fact(n):
  if n==0 or n==1:
    return 1
  else:
    return n*fact(n-1)
def sf(n):
  if n==0 or n==1:
    return 1
  else:
    return fact(n)*sf(n-1)
print(sf(4))
print(fact(4))
```
